```python
target_cities = ["Mumbai", "Navi Mumbai", "Thane"]

for city in target_cities:
    count = df["City"].str.contains(city, case=False, na=False).sum()
    print(f"{city}: {count}")
```

```python
for keyword in ["Mumbai", "Navi Mumbai", "Thane"]:
    mask = df["Locality"].str.contains(keyword, case=False, na=False)
    print(f"\n{keyword}: {mask.sum()} restaurants")
    print(df.loc[mask, "Locality"].value_counts().head(15))
```

```python
mumbai_df = df[df["City"].str.contains("Mumbai", case=False, na=False)]

print("Mumbai restaurants:", len(mumbai_df))
```
